## Stratégie provisoire de traitement des valeurs manquantes

- `Type2` : absence souvent structurelle correspondant à un Pokémon monotype.
- `Experience` : valeur liée à la courbe de croissance ; imputation possible pour certaines formes temporaires à partir de la forme de référence.
- `TauxCapture` : certaines formes temporaires ne sont pas capturables en tant que telles ; les valeurs manquantes peuvent donc être conservées.
- `Poids_kg` : les formes Gigamax peuvent ne pas disposer d'un poids officiel ; aucune imputation arbitraire.
- `Type1` : valeurs réellement manquantes, concentrées sur des Pokémon à formes multiples ; complétion manuelle à partir des sources de référence.
- Statistiques de combat : étude séparée des cas manquants avant décision.


### Interprétation des valeurs manquantes selon les formes spéciales

Les valeurs manquantes ne sont pas réparties aléatoirement dans le jeu de données.

Les formes Gigamax concentrent notamment les valeurs manquantes concernant l'expérience, le taux de capture, le poids et les statistiques de combat. Cela suggère des absences structurelles liées à la nature de ces formes et aux informations disponibles dans la source.

Les Méga-évolutions présentent également de nombreuses valeurs manquantes pour l'expérience et le taux de capture, alors que leur poids et leur type principal restent renseignés.

Ces résultats montrent qu'une suppression globale des observations incomplètes introduirait un biais important en éliminant une grande partie des formes spéciales.

### Analyse des doublons et des formes multiples

Le jeu de données contient 1241 observations pour 904 numéros Pokédex distincts. Une ligne ne représente donc pas nécessairement une espèce Pokémon unique : plusieurs formes d'un même Pokémon peuvent être présentes.

131 lignes sont détectées comme doublons exacts. Ces doublons ne peuvent cependant pas être supprimés automatiquement.

Plusieurs situations ont été identifiées :

- certaines lignes correspondent à de véritables formes différentes d'un même Pokémon ;
- certaines apparaissent artificiellement identiques à cause de valeurs manquantes, notamment pour les types ;
- certaines correspondent à des formes essentiellement cosmétiques partageant l'ensemble des variables analysées.

La déduplication éventuelle sera donc effectuée après le traitement des données manquantes afin d'éviter la suppression de formes réellement distinctes.


Le maintien de nombreuses formes cosmétiques strictement identiques pourrait également surpondérer certains Pokémon dans les analyses factorielles et la classification.

```python
duplicate_groups = (
    doublons_exacts
    .groupby(["Numero", "Nom"])
    .size()
    .sort_values(ascending=False)
)

duplicate_groups.head(30)
```

## Conclusion de l'audit

L'audit met en évidence plusieurs spécificités importantes du jeu de données :

- plusieurs formes d'un même Pokémon peuvent partager le même numéro Pokédex ;
- certaines formes sont réellement différentes sur le plan statistique ou typologique ;
- d'autres formes sont essentiellement cosmétiques et peuvent apparaître comme des doublons exacts ;
- les valeurs manquantes sont fortement structurées et liées à certaines formes particulières ;
- une suppression naïve des doublons ou des valeurs manquantes introduirait donc un biais important.

Le nettoyage sera réalisé de manière ciblée afin de préserver les formes réellement distinctes tout en évitant la surreprésentation de formes cosmétiques identiques.
